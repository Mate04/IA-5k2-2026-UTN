La meta es clasificar a los clientes en dos categorías:

1.  **Churn (Abandono):** Si el cliente es propenso a abandonar el servicio (Verdadero - Clase 1).
2.  **No churn (No Abandono):** Si el cliente es fiel al servicio (Falso - Clase 0).

La empresa proporciona un pequeño conjunto de datos de ejemplo para desarrollar un modelo predictivo que pueda clasificar correctamente a los clientes.

c) graficar la frontera entre las clases e indicar los parámetros (pesos) que la definen

El objetivo de este punto es encontrar y graficar una línea recta (frontera de decisión) que separe a los clientes entre las dos clases ("Churn" y "No Churn") y especificar los parámetros o pesos que definen esa línea.

Para lograr esto, seguiremos estos pasos:

1. Visualizar los Datos: Primero, graficamos los datos proporcionados en un plano 2D. El eje X representará el "Número de quejas" (x1) y el eje Y representará el "Porcentaje de uso" (x2).

•	Clase 0 (No Churn): (0, 100), (1, 80), (0, 90), (1, 70).

•	Clase 1 (Churn): (2, 30), (3, 20), (2, 50), (4, 10).

Al observar el gráfico, podemos notar que los grupos están claramente separados. Todos los puntos de la clase "No Churn" se encuentran a la izquierda y los de la clase "Churn" a la derecha.

2. Determinar la Frontera de Decisión

Un modelo de clasificador lineal, como el Perceptrón, crea una frontera que es una línea recta para separar las clases. La ecuación general de una recta en este plano es: w1x1+w2x2+b=0

donde:

•	w1 y w2 son los pesos (parámetros) que indican la inclinación de la línea.

•	x1 y x2 son las variables de entrada (quejas y uso).

•	b es el sesgo (bias), que desplaza la línea.

Al colocar los puntos en el grafico se puede obervar que los datos se pueden separar:

1. Con una línea vertical. Todos los clientes que no abandonan el servicio tienen 1 queja o menos, mientras que los que sí lo hacen tienen 2 o más.
Por lo tanto, podemos trazar una línea vertical justo en medio de estos dos grupos, por ejemplo, en x1=1.5.

La ecuación de esta línea es: x1=1.5

Para que coincida con la forma general, la reescribimos como: 1⋅x1+0⋅x2−1.5=0

2. Con una línea en diagonal. Como los parámetros no se pueden calcular a mano, sino que son el resultado del entrenamiento del modelo con los datos, se puede usar algunos valores estimados, que logren visualizar la recta que separa las dos clases.

3. Indicar los Parámetros (Pesos)

Si usamos el primer gráfico, más intuitivo y simple. Al comparar la ecuación de nuestra línea con la forma general w1x1+w2x2+b=0, podemos identificar directamente los parámetros que la definen:

•	Peso para "Número de quejas" (w1): 1

•	Peso para "Porcentaje de uso" (w2): 0 (Esto tiene sentido, ya que la clasificación en este modelo simple no depende del porcentaje de uso, solo del número de quejas).

•	Sesgo (b): -1.5

Estos parámetros definen un clasificador que predice "Churn" si x1−1.5≥0 y "No Churn" si x1−1.5<0.

4. Graficar la Frontera entre las Clases

Si usamos el primer gráfico, dibujamos la línea x1=1.5 en la gráfica junto con los puntos de datos para visualizar cómo separa las dos clases.

Como se puede ver, la línea vertical divide el espacio en dos regiones, clasificando correctamente a todos los clientes del conjunto de datos proporcionado.

Se observa que los datos se pueden separar perfectamente con una línea recta. Todos los clientes que no abandonan el servicio tienen 1 queja o menos, mientras que los que sí lo hacen tienen 2 o más.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Perceptron

# Datos: X = [número de quejas, porcentaje de uso], Y = 1 (Churn) / 0 (No Churn)
X = np.array([[0, 100], [2, 30], [1, 80], [3, 20], [0, 90], [2, 50], [1, 70], [4, 10]])
Y = np.array([0, 1, 0, 1, 0, 1, 0, 1])

# Entrenamiento del perceptrón (tol=None para que no corte antes de separar todo)
perceptron = Perceptron(max_iter=1000, tol=None, random_state=0)
perceptron.fit(X, Y)

w1, w2 = perceptron.coef_[0]
b = perceptron.intercept_[0]
print(f"Pesos: w1 = {w1}, w2 = {w2}, bias b = {b}")
print(f"Frontera de decisión: {w1}·x1 + ({w2})·x2 + ({b}) = 0")
print("Exactitud sobre los datos:", perceptron.score(X, Y))

# Función de decisión: f(x) = w1·x1 + w2·x2 + b  ->  clase 1 si f(x) > 0, clase 0 si no
def funcion_decision(x1, x2):
    return w1 * x1 + w2 * x2 + b

for (x1, x2), y in zip(X, Y):
    f = funcion_decision(x1, x2)
    print(f"x = [{x1}, {x2}]  f(x) = {f:8.1f}  -> predice {int(f > 0)}  (real {y})")

# Gráfico: regiones de decisión + frontera + puntos
xx1, xx2 = np.meshgrid(np.linspace(-0.5, 4.5, 300), np.linspace(0, 110, 300))
zz = perceptron.predict(np.c_[xx1.ravel(), xx2.ravel()]).reshape(xx1.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx1, xx2, zz, alpha=0.2, cmap="bwr")

# Recta frontera: x2 = -(w1·x1 + b) / w2
x1_linea = np.linspace(-0.5, 4.5, 100)
x2_linea = -(w1 * x1_linea + b) / w2
plt.plot(x1_linea, x2_linea, "k--", label="Frontera de decisión")

plt.scatter(X[Y == 0, 0], X[Y == 0, 1], c="blue", s=80, edgecolors="k", label="No Churn (0)")
plt.scatter(X[Y == 1, 0], X[Y == 1, 1], c="red", s=80, edgecolors="k", label="Churn (1)")

plt.xlim(-0.5, 4.5)
plt.ylim(0, 110)
plt.xlabel("Número de quejas (x1)")
plt.ylabel("Porcentaje de uso (x2)")
plt.title("Perceptrón - Frontera de decisión")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression

# Regresión logística: modela P(Churn | x) = sigmoide(w1·x1 + w2·x2 + b)
logistica = LogisticRegression(max_iter=10000)
logistica.fit(X, Y)

w1_lr, w2_lr = logistica.coef_[0]
b_lr = logistica.intercept_[0]
print(f"Pesos: w1 = {w1_lr:.4f}, w2 = {w2_lr:.4f}, bias b = {b_lr:.4f}")
print(f"Frontera de decisión (P = 0.5): {w1_lr:.4f}·x1 + ({w2_lr:.4f})·x2 + ({b_lr:.4f}) = 0")
print("Exactitud sobre los datos:", logistica.score(X, Y))

# A diferencia del perceptrón, da una probabilidad para cada cliente
probs = logistica.predict_proba(X)[:, 1]
for (x1, x2), y, p in zip(X, Y, probs):
    print(f"x = [{x1}, {x2}]  P(Churn) = {p:.3f}  -> predice {int(p > 0.5)}  (real {y})")

# Gráfico: probabilidad de Churn en todo el plano + frontera P = 0.5
xx1, xx2 = np.meshgrid(np.linspace(-0.5, 4.5, 300), np.linspace(0, 110, 300))
pp = logistica.predict_proba(np.c_[xx1.ravel(), xx2.ravel()])[:, 1].reshape(xx1.shape)

plt.figure(figsize=(8, 6))
cf = plt.contourf(xx1, xx2, pp, levels=20, alpha=0.4, cmap="bwr")
plt.colorbar(cf, label="P(Churn)")

x1_linea = np.linspace(-0.5, 4.5, 100)
x2_linea = -(w1_lr * x1_linea + b_lr) / w2_lr
plt.plot(x1_linea, x2_linea, "k--", label="Frontera (P = 0.5)")

plt.scatter(X[Y == 0, 0], X[Y == 0, 1], c="blue", s=80, edgecolors="k", label="No Churn (0)")
plt.scatter(X[Y == 1, 0], X[Y == 1, 1], c="red", s=80, edgecolors="k", label="Churn (1)")

plt.xlim(-0.5, 4.5)
plt.ylim(0, 110)
plt.xlabel("Número de quejas (x1)")
plt.ylabel("Porcentaje de uso (x2)")
plt.title("Regresión logística - Frontera de decisión")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

```python
from sklearn.svm import SVC

# SVM lineal: busca la recta que deja el MAYOR margen posible entre las clases
svm = SVC(kernel="linear", C=1.0)
svm.fit(X, Y)

w1_svm, w2_svm = svm.coef_[0]
b_svm = svm.intercept_[0]
margen = 2 / np.linalg.norm(svm.coef_[0])
print(f"Pesos: w1 = {w1_svm:.4f}, w2 = {w2_svm:.4f}, bias b = {b_svm:.4f}")
print(f"Frontera de decisión: {w1_svm:.4f}·x1 + ({w2_svm:.4f})·x2 + ({b_svm:.4f}) = 0")
print("Exactitud sobre los datos:", svm.score(X, Y))
print("Vectores de soporte:\n", svm.support_vectors_)
print(f"Ancho del margen: {margen:.2f}")

# Gráfico: regiones + frontera + márgenes (f(x) = ±1) + vectores de soporte
xx1, xx2 = np.meshgrid(np.linspace(-0.5, 4.5, 300), np.linspace(0, 110, 300))
zz = svm.predict(np.c_[xx1.ravel(), xx2.ravel()]).reshape(xx1.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx1, xx2, zz, alpha=0.2, cmap="bwr")

x1_linea = np.linspace(-0.5, 4.5, 100)
for nivel, estilo, etiqueta in [(0, "k--", "Frontera de decisión"), (1, "k:", "Márgenes"), (-1, "k:", None)]:
    # w1·x1 + w2·x2 + b = nivel  ->  x2 = (nivel - w1·x1 - b) / w2
    plt.plot(x1_linea, (nivel - w1_svm * x1_linea - b_svm) / w2_svm, estilo, label=etiqueta)

plt.scatter(X[Y == 0, 0], X[Y == 0, 1], c="blue", s=80, edgecolors="k", label="No Churn (0)")
plt.scatter(X[Y == 1, 0], X[Y == 1, 1], c="red", s=80, edgecolors="k", label="Churn (1)")
plt.scatter(svm.support_vectors_[:, 0], svm.support_vectors_[:, 1], s=250,
            facecolors="none", edgecolors="green", linewidths=2, label="Vectores de soporte")

plt.xlim(-0.5, 4.5)
plt.ylim(0, 110)
plt.xlabel("Número de quejas (x1)")
plt.ylabel("Porcentaje de uso (x2)")
plt.title("SVM lineal - Frontera de decisión y margen")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

```python
# Comparación de los tres modelos
modelos = {
    "Perceptrón": perceptron,
    "Regresión logística": logistica,
    "SVM lineal": svm,
}
# (la logística y la SVM quedan casi superpuestas, por eso se dibujan con distinto grosor y trazo)
estilos = {
    "Perceptrón": dict(color="purple", lw=2),
    "Regresión logística": dict(color="orange", lw=5),
    "SVM lineal": dict(color="green", lw=2, ls="--"),
}

# Los pesos de cada modelo tienen escalas distintas (la recta w·x + b = 0 es la misma si se
# multiplican todos por una constante), así que para comparar se escribe cada frontera
# como x2 = pendiente·x1 + ordenada
print(f"{'Modelo':<22}{'w1':>10}{'w2':>10}{'b':>10}{'pendiente':>11}{'ordenada':>10}{'exactitud':>11}")
for nombre, m in modelos.items():
    w1_m, w2_m = m.coef_[0]
    b_m = m.intercept_[0]
    print(f"{nombre:<22}{w1_m:>10.4f}{w2_m:>10.4f}{b_m:>10.4f}"
          f"{-w1_m / w2_m:>11.2f}{-b_m / w2_m:>10.2f}{m.score(X, Y):>11.2f}")

# Distancia mínima de los puntos a cada frontera (en unidades de los datos):
# cuanto más grande, más "segura" es la separación
print()
for nombre, m in modelos.items():
    dist = np.abs(m.decision_function(X)) / np.linalg.norm(m.coef_[0])
    print(f"{nombre:<22} distancia mínima a la frontera: {dist.min():.2f}")

# Predicción sobre clientes nuevos (no vistos)
nuevos = np.array([[1, 40], [2, 65], [4, 70], [0, 50]])
print("\nPredicciones sobre clientes nuevos [quejas, uso]:")
for nombre, m in modelos.items():
    print(f"{nombre:<22}", m.predict(nuevos))

# Gráfico: las tres fronteras juntas
plt.figure(figsize=(8, 6))
x1_linea = np.linspace(-0.5, 4.5, 100)
for nombre, m in modelos.items():
    w1_m, w2_m = m.coef_[0]
    b_m = m.intercept_[0]
    plt.plot(x1_linea, -(w1_m * x1_linea + b_m) / w2_m, label=nombre, **estilos[nombre])

plt.scatter(X[Y == 0, 0], X[Y == 0, 1], c="blue", s=80, edgecolors="k", label="No Churn (0)")
plt.scatter(X[Y == 1, 0], X[Y == 1, 1], c="red", s=80, edgecolors="k", label="Churn (1)")
plt.scatter(nuevos[:, 0], nuevos[:, 1], c="gray", marker="x", s=100, label="Clientes nuevos")

plt.xlim(-0.5, 4.5)
plt.ylim(0, 110)
plt.xlabel("Número de quejas (x1)")
plt.ylabel("Porcentaje de uso (x2)")
plt.title("Comparación de fronteras de decisión")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

**Comparación de los tres modelos**

Los tres modelos separan perfectamente los 8 clientes (exactitud 1.0), pero encuentran fronteras distintas:

| | Perceptrón | Regresión logística | SVM lineal |
|---|---|---|---|
| Frontera encontrada | recta inclinada (x2 ≈ 25.5·x1 + 7.1) | casi horizontal (x2 ≈ 60) | casi horizontal (x2 ≈ 60) |
| Criterio | se detiene en la **primera** recta que separa sin errores | maximiza la **verosimilitud** (probabilidades) | maximiza el **margen** entre las clases |
| Salida | clase (0/1) | probabilidad P(Churn) + clase | clase (0/1) + distancia a la frontera |
| Solución única | no (depende del orden de los datos y la inicialización) | sí | sí |
| Distancia mínima de un punto a la frontera | ≈ 0.3 (pasa pegada al cliente [2, 50]) | ≈ 10 | ≈ 10 (máxima posible) |

Conclusiones:

* El **perceptrón** encuentra *una* solución válida, pero no necesariamente la mejor: su recta pasa muy cerca de un punto, así que un cliente nuevo un poco distinto podría quedar mal clasificado.
* La **SVM** elige la recta que queda "en el medio" de las dos clases, definida sólo por los **vectores de soporte** ([1, 70] y [2, 50]). Es la separación más robusta.
* La **regresión logística** llega prácticamente a la misma frontera que la SVM (con datos separables tiende a ella), y además da la **probabilidad** de abandono, útil para priorizar a qué clientes contactar.
* Como los datos no están escalados, el porcentaje de uso (0-100) domina sobre el número de quejas (0-4), por eso la logística y la SVM quedan casi horizontales. Con clientes nuevos como [0, 50] o [4, 70] los modelos **no coinciden**: con tan pocos datos, la frontera elegida cambia la predicción.

d) Tipo de ambiente en que se desempeñaría un agente que use el modelo planteado — indicar y justificar cada dimensión

1.	Totalmente observable / Parcialmente observable:

Parcialmente observable. Justificación: el modelo usa sólo dos atributos (quejas y % uso). En la práctica existen otros factores (satisfacción, precio, competencia, soporte) que no están observados; por tanto el agente tiene observación parcial del estado real del cliente.

2.	Determinístico / Estocástico:

Estocástico. Justificación: la relación entre características observadas y abandono no es determinística: dos clientes con las mismas características pueden comportarse distinto. Por lo tanto, el resultado no es determinista.

3.	Episódico / Secuencial:

Episódico (en la forma planteada). Justificación: cada predicción de churn aquí se hace por cliente con sus características actuales (instancia independiente). Cada predicción de cliente es un evento independiente. Clasificar a un cliente no influye en la clasificación del siguiente. El agente no necesita recordar acciones pasadas. Si se considerara evolución temporal (monitoreo a lo largo de meses) sería secuencial, pero el enunciado plantea una clasificación de instances independientes.

4.	Estático / Dinámico:

Estático (en la forma planteada). Justificación: el modelo decide con las características actuales; el entorno no cambia durante la decisión. Si el entorno temporalmente cambia tras la acción (p. ej. contacto comercial), sería dinámico.

5.	Discreto / Continuo:

Discreto (salida). Las acciones del agente (clasificar como Churn o No Churn) son discretas. Las entradas son una mezcla (quejas es discreto, uso es continuo pero se presenta como entero), pero el espacio de estados y acciones es fundamentalmente discreto. Justificación: la acción de clasificar es discreta (Si/No). Las características incluyen continuas (% uso) y discretas (nº quejas).

6.	Individual / Multi-agente:

Individual. Justificación: el agente examina cada cliente individualmente y decide su clasificación. El agente clasificador opera de forma individual y no interactúa ni compite con otros agentes.

7.	Conocido / Desconocido (sobre el entorno):

Desconocido. Justificación: El agente no conoce las reglas exactas que determinan si un cliente abandonará el servicio. Debe aprender estas reglas a partir de los datos (por eso aprendemos un modelo).

Usando el enunciado del ejercicio 1 y suponiendo que, luego de aplicar algún modelo con 200 datos de prueba, se obtienen los siguientes resultados:
TP(VP) = 70 (verdaderos positivos), TN(VN) = 40 (verdaderos negativos), FP(FP) = 30 (falsos positivos), FN(FN) = 60 (falsos negativos)

Se pide:

a) Graficar e interpretar la matriz de confusión. (5%)

b) Si el clasificador fuera perfecto y teniendo en cuenta el punto anterior (matriz de confusión), ¿Cómo quedaría la matriz de confusión?. Escribir los valores.

c) Según los valores anteriores, interpretar las métricas de Precisión, Exactitud (Accuracy), Sensibilidad (Recall) y Especificidad (Specificity), para el modelo dado.

## Resolución

a) Matriz de confusión

|                | Predicción: Positivo | Predicción: Negativo |
|---------------:|:---------------------:|:---------------------:|
| **Real: Positivo** | TP (VP) = 70         | FN (FN) = 60         |
| **Real: Negativo** | FP (FP) = 30         | TN (VN) = 40         |

**Interpretación:**

*   **Verdaderos Positivos (TP/VP):** 70 casos donde el modelo predijo positivo y el valor real también era positivo.
*   **Verdaderos Negativos (TN/VN):** 40 casos donde el modelo predijo negativo y el valor real también era negativo.
*   **Falsos Positivos (FP):** 30 casos donde el modelo predijo positivo, pero el valor real era negativo (Error de Tipo I).
*   **Falsos Negativos (FN):** 60 casos donde el modelo predijo negativo, pero el valor real era positivo (Error de Tipo II).

b) Matriz de Confusión para un Clasificador Perfecto

Un clasificador perfecto no comete errores, por lo tanto, los Falsos Positivos (FP) y Falsos Negativos (FN) serían cero.

•	Total de clientes "No Churn" reales: TN + FP = 40 + 30 = 70.

•	Total de clientes "Churn" reales: TP + FN = 70 + 60 = 130.

La matriz perfecta quedaría así:

|                | Predicción: Positivo | Predicción: Negativo |
|---------------:|:---------------------:|:---------------------:|
| **Real: Positivo** | TP (VP) = 70         | FN (FN) = 0         |
| **Real: Negativo** | FP (FP) = 0         | TN (VN) = 130         |



c) Usando los valores de la matriz de confusión:

    TP = 70
    TN = 40
    FP = 30
    FN = 60

**Precisión (Precision):**
* Formula: $VP / (VP + FP)$
* Calculo: $70 / (70 + 30) = 70 / 100 = 0.7$
* Interpretación: De todas las instancias predichas como positivas, el 70% fueron realmente positivas.

**Exactitud (Accuracy):**
* Formula: $(VP + VN) / (VP + FP + FN + VN)$
* Cálculo: $(70 + 40) / (70 + 30 + 60 + 40) = 110 / 200 = 0.55$
* Interpretación: El modelo clasificó correctamente el 55% del total de instancias.

**Sensibilidad (Recall):**
* Formula: $VP / (VP + FN)$
* Cálculo: $70 / (70 + 60) = 70 / 130 \approx 0.538$
* Interpretación: De todas las instancias positivas reales, el modelo identificó correctamente aproximadamente el 53.8%.

**Especificidad (Specificity):**
* Formula: $VN / (VN + FP)$
* Cálculo: $40 / (40 + 30) = 40 / 70 \approx 0.571$
* Interpretación: De todas las instancias negativas reales, el modelo identificó correctamente aproximadamente el 57.1%.

Comentario global sobre las métricas:

•	Precisión relativamente buena (70%) —cuando predice abandono tiene probabilidad razonable de estar en lo cierto— pero recall es bajo-moderado (≈54%), es decir, el sistema deja escapar muchos abandonos reales (riesgo comercial).

## Ejercicio 4

Usando los datos del ejercicio 1, ¿Cómo serían dos pasos de entrenamiento utilizando un perceptrón simple, con un bias de 1 y una taza de aprendizaje ($\alpha$) de 0.3?

## Ejemplo de Entrenamiento con un Perceptrón Simple

Vamos a demostrar dos pasos de entrenamiento utilizando un perceptrón simple con los siguientes parámetros:

*   **Bias inicial:** 1
*   **Tasa de aprendizaje ($\alpha$):** 0.3

Usaremos los dos primeros ejemplos del conjunto de datos para la demostración.

**Datos de entrenamiento:**

| Número de quejas | Porcentaje de uso (%) | Propenso a abandonar | Vector de Entrada ($x$) | Vector de Salida deseada ($y$) |
|:---:|:---:|:---:|:---:|:---:|
| 0 | 100 | No | [0, 100] | 0 |
| 2 | 30 | Si | [2, 30] | 1 |

**Inicialización de pesos:**

Comenzaremos con pesos iniciales aleatorios (o podríamos inicializarlos a cero por simplicidad en este ejemplo, pero usaremos valores pequeños aleatorios para una demostración más realista).

$w = [w_1, w_2]$

$b = 1$

Pesos iniciales (aleatorios pequeños): $w = [0.1, -0.05]$

>
> _Consideramos el vector ampliado donde el bias $b$ es representado por $w_3$_
>

Pesos iniciales (aleatorios pequeños): $w = [0.1, -0.05, 1]$

### Paso 1: Entrenamiento con el primer ejemplo

**Entrada ($x$):** $[0, 100, 1]$

**Salida deseada ($y$):** 0

1.  **Calcular la entrada neta ($z$):**
    
    $z = w_1 \cdot x_1 + w_2 \cdot x_2 + w_3$
    
    $z = (0.1 \cdot 0) + (-0.05 \cdot 100) + 1$
    
    $z = 0 - 5 + 1 = -4$

2.  **Aplicar la función de activación (escalón):**
    
    Para un perceptrón simple, usamos una función escalón. Si $z \ge 0$, la salida es 1; si $z < 0$, la salida es 0.
    
    Salida predicha ($\hat{y}$): Como $z = -4 < 0$, $\hat{y} = 0$.

3.  **Calcular el error:**
    
    Error = $y - \hat{y}$
    
    Error = $0 - 0 = 0$

4.  **Actualizar pesos y bias:**
    
    Como el error es 0, no hay actualización de pesos ni bias en este paso. Los pesos y bias se mantienen igual.
    
    $w_{\text{nuevo}} = w_{\text{viejo}}$

    Pesos después del paso 1 (incluyendo $w_3$): $w = [0.1, -0.05, 1]$
    
### Paso 2: Entrenamiento con el segundo ejemplo

**Entrada ($x$):** $[2, 30, 1]$

**Salida deseada ($y$):** 1

1.  **Calcular la entrada neta ($z$):**
    
    Usamos los pesos y bias actualizados (que en este caso no cambiaron).
    
    $z = w_1 \cdot x_1 + w_2 \cdot x_2 + w_3$
    
    $z = (0.1 \cdot 2) + (-0.05 \cdot 30) + 1$
    
    $z = 0.2 - 1.5 + 1 = -0.3$

2.  **Aplicar la función de activación (escalón):**
    
    Salida predicha ($\hat{y}$): Como $z = -0.3 < 0$, $\hat{y} = 0$.

3.  **Calcular el error:**
    
    Error = $y - \hat{y}$
    
    Error = $1 - 0 = 1$

4.  **Actualizar pesos y bias:**
    
    Como el error es 1, actualizamos los pesos y el bias usando la regla de actualización del perceptrón:
    
    $w_{i, \text{nuevo}} = w_{i, \text{viejo}} + \alpha \cdot \text{Error} \cdot x_i$

    Actualización del peso $w_1$:
    
    $w_{1, \text{nuevo}} = w_{1, \text{viejo}} + \alpha \cdot \text{Error} \cdot x_1$
    
    $w_{1, \text{nuevo}} = 0.1 + 0.3 \cdot 1 \cdot 2$
    
    $w_{1, \text{nuevo}} = 0.1 + 0.6 = 0.7$

    Actualización del peso $w_2$:
    
    $w_{2, \text{nuevo}} = w_{2, \text{viejo}} + \alpha \cdot \text{Error} \cdot x_2$
    
    $w_{2, \text{nuevo}} = -0.05 + 0.3 \cdot 1 \cdot 30$
    
    $w_{2, \text{nuevo}} = -0.05 + 9 = 8.95$

    Actualización del peso $w_3$ (bias $b$):
    
    $w_{\text{3,nuevo}} = w_{\text{3,viejo}} + \alpha \cdot \text{Error}$
    
    $w_{\text{3,nuevo}} = 1 + 0.3 \cdot 1$
    
    $w_{\text{3,nuevo}} = 1 + 0.3 = 1.3$

**Resultados después de dos pasos de entrenamiento:**

Pesos: $w = [0.7, 8.95, 1.3]$

Este es un ejemplo simple de cómo el perceptrón ajusta sus pesos y bias en cada paso de entrenamiento basado en el error para aprender a clasificar los datos.

Con más iteraciones y datos, el perceptrón (si el problema es linealmente separable) convergerá a un conjunto de pesos y bias que definen la frontera de decisión.
